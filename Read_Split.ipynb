## This notebook loads the HT-CC dataset, converts complex I/Q frames into a real-valued two-channel format, assigns class and SNR labels, and creates a reproducible train/test split.


```python
import numpy as np 
import matplotlib.pyplot as plt 
import os
import scipy
import random
from sklearn.model_selection import train_test_split
import time
```

```python
SEED = 0
```

```python
Dir = "./covert_channel"   # Change this to your dataset directory
```

## Data Preprocessing

```python
data_HT0 = None
data_HT1 = None
data_HT2 = None
data_HT3 = None
data_HT4 = None
data_HT5 = None



k1 = 'rxSigFull'
frame_size = 640
for dirname, dirs, filenames in os.walk(Dir):
    for filename in sorted(filenames):
        f=(os.path.join(dirname, filename))
        if ("HT0" in f):
            print(f)
            fI0 = scipy.io.loadmat(f)
            data_HT0 = (np.concatenate((data_HT0,fI0[k1]),axis = 0) if (data_HT0 is not None) else fI0[k1])
        elif("HT1" in f):
            print(f)
            fI1 = scipy.io.loadmat(f)
            data_HT1 = (np.concatenate((data_HT1,fI1[k1]),axis = 0) if (data_HT1 is not None) else fI1[k1])
        elif ("HT2" in f):
            print(f)
            fI2 = scipy.io.loadmat(f)
            data_HT2 = (np.concatenate((data_HT2,fI2[k1]),axis = 0) if (data_HT2 is not None) else fI2[k1])
        elif ("HT3" in f):
            print(f)
            fI3 = scipy.io.loadmat(f)
            data_HT3 = (np.concatenate((data_HT3,fI3[k1]),axis = 0) if (data_HT3 is not None) else fI3[k1])
        elif("HT4" in f):
            print(f)
            fI4 = scipy.io.loadmat(f)
            data_HT4 = (np.concatenate((data_HT4,fI4[k1]),axis = 0) if (data_HT4 is not None) else fI4[k1])
```

```python
print(np.shape(data_HT0))
print(np.shape(data_HT1))
print(np.shape(data_HT2))
print(np.shape(data_HT3))
print(np.shape(data_HT4))
```

```python
###################
#Framing
##################
data_HT0_reshaped = np.reshape(data_HT0[0:frame_size*int(np.shape(data_HT0)[0]/frame_size)],(1,-1,frame_size))
data_HT1_reshaped = np.reshape(data_HT1[0:frame_size*int(np.shape(data_HT1)[0]/frame_size)],(1,-1,frame_size))
data_HT2_reshaped = np.reshape(data_HT2[0:frame_size*int(np.shape(data_HT2)[0]/frame_size)],(1,-1,frame_size))
data_HT3_reshaped = np.reshape(data_HT3[0:frame_size*int(np.shape(data_HT3)[0]/frame_size)],(1,-1,frame_size))
data_HT4_reshaped = np.reshape(data_HT4[0:frame_size*int(np.shape(data_HT4)[0]/frame_size)],(1,-1,frame_size))

###################
#Labeling
##################
y_HT0 = np.zeros(np.shape(data_HT0_reshaped)[1])
y_HT1 = np.ones(np.shape(data_HT1_reshaped)[1])
y_HT2 = np.empty(np.shape(data_HT2_reshaped)[1])
y_HT2.fill(2)
y_HT3 = np.empty(np.shape(data_HT3_reshaped)[1])
y_HT3.fill(3)
y_HT4 = np.empty(np.shape(data_HT4_reshaped)[1])
y_HT4.fill(4)

SNR_LABEL_LENGTH = int(np.shape(data_HT1_reshaped)[1]/8)

y_SNR1 = np.ones(SNR_LABEL_LENGTH)
y_SNR2 = np.empty(SNR_LABEL_LENGTH)
y_SNR2.fill(2)
y_SNR3 = np.empty(SNR_LABEL_LENGTH)
y_SNR3.fill(3)
y_SNR4 = np.empty(SNR_LABEL_LENGTH)
y_SNR4.fill(4)
y_SNR5 = np.empty(SNR_LABEL_LENGTH)
y_SNR5.fill(5)
y_SNR6 = np.empty(SNR_LABEL_LENGTH)
y_SNR6.fill(6)
y_SNR7 = np.empty(SNR_LABEL_LENGTH)
y_SNR7.fill(7)
y_SNR8 = np.empty(SNR_LABEL_LENGTH)
y_SNR8.fill(8)

print(np.shape(y_SNR1))
y_SNR = np.concatenate((y_SNR1,y_SNR2,y_SNR3,y_SNR4,y_SNR5,y_SNR6,y_SNR7,y_SNR8),axis = 0)
Y_SNR = np.reshape(np.concatenate((y_SNR,y_SNR,y_SNR,y_SNR,y_SNR),axis = 0),(-1,1))

print("y_SNR Length",np.shape(Y_SNR))

print(np.shape(data_HT0_reshaped))
print(np.shape(y_HT0))
print(np.shape(data_HT1_reshaped))
print(np.shape(y_HT1))
print(np.shape(data_HT2_reshaped))
print(np.shape(y_HT2))
print(np.shape(data_HT3_reshaped))
print(np.shape(y_HT3))
print(np.shape(data_HT4_reshaped))
print(np.shape(y_HT4))
```

```python
plt.figure(1)
plt.grid()
plt.plot((data_HT0_reshaped[0,1,:]))
```

```python
#################################
#IQ pre processing NON infected
#################################
data_HT0_IQ = np.concatenate((np.real(data_HT0_reshaped),np.imag(data_HT0_reshaped)), axis = 0)
data_HT1_IQ = np.concatenate((np.real(data_HT1_reshaped),np.imag(data_HT1_reshaped)), axis = 0)
data_HT2_IQ = np.concatenate((np.real(data_HT2_reshaped),np.imag(data_HT2_reshaped)), axis = 0)
data_HT3_IQ = np.concatenate((np.real(data_HT3_reshaped),np.imag(data_HT3_reshaped)), axis = 0)
data_HT4_IQ = np.concatenate((np.real(data_HT4_reshaped),np.imag(data_HT4_reshaped)), axis = 0)

print(np.shape(data_HT0_IQ))
print(np.shape(data_HT1_IQ))
print(np.shape(data_HT2_IQ))
print(np.shape(data_HT3_IQ))
print(np.shape(data_HT4_IQ))
```

```python
X = np.concatenate((data_HT0_IQ,data_HT1_IQ,data_HT2_IQ,data_HT3_IQ,data_HT4_IQ) , axis = 1)
X = np.swapaxes(X,0,1)
Y = np.reshape(np.concatenate((y_HT0,y_HT1,y_HT2,y_HT3,y_HT4) , axis = 0),(-1,1))
print(np.shape(X))
print(np.shape(Y))
```

```python
X = X.reshape(X.shape[0], 1, 2, X.shape[2])
print(np.shape(X))
print(np.shape(Y))
```

```python
X_train, X_test, y_train, y_test, idx_train, idx_test = train_test_split(X, Y, np.arange(np.shape(X)[0]),test_size=0.3, shuffle = True, random_state=SEED)
```
